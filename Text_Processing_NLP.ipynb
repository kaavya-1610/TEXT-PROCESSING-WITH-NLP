```python
'''
1. Lowercasing

2. Removing punctuation

3. Stopwords removal

4. Tokenization

Stemming - It reduces words to root form by applying simple rules(cutting prefix and suffix),
           but it may not be valid or meaningful.
e.g.: Learning - Learn
      Happily - Happi
      better - bett
                              (or)

Lemmatization - Find root words, correct dictionary form (lemma).
e.g.: Learning - Learn
      Happily - Happy(because it directly connected with a dictionary (lexicon).
Use this if meaning is necessary'''
```

```python
text = "Data Science is an emerging field. That's True! Is it? Yes. I heard it from my professor."
```

```python
import re
import nltk

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
```

```python
# Remove all the punctuations
text = re.sub(r'[^\w\s]','',text)
text
```

```python
# Lowercase
lowercase = text.lower()
lowercase
```

```python
# Tokenize
tokens = word_tokenize(lowercase)
tokens
```

```python
# Stop words
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words]
filtered_tokens
```

```python
# Stemming

stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
stemmed_tokens
```

```python
# Lemmatization

lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
lemmatized_tokens
```
